```python
# comnining text
df['combined_text'] = (
    df['title'] + " " +
    df['description'] + " " +
    df['input_description'] + " " +
    df['output_description']
)
```

### Text cleaning

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

df['clean_text'] = df['combined_text'].apply(clean_text)
```

```python
df[['clean_text', 'problem_class', 'problem_score']].head()
```

```python
clean_df = df[
    [
        'title',
        'description',
        'input_description',
        'output_description',
        'clean_text',
        'problem_class',
        'problem_score'
    ]
]
```
